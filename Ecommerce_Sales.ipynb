```python
#Remove Only Delivered Orders
orders = orders.drop(
    orders[
        (orders["order_status"] == "delivered") &
        (orders["order_approved_at"].isnull())
    ].index
)
```

```python
orders["order_approved_at"].isnull().sum()
```

### Handling Missing Values in `order_approved_at`

The `order_approved_at` column contained **160 missing values**.

To understand the reason for these missing values, the corresponding `order_status` values were examined.

The analysis showed:

- **141** canceled orders
- **14** delivered orders
- **5** created orders

Missing approval timestamps for **canceled** and **created** orders were retained because they represent valid business scenarios where an order may not have been approved.

However, the **14 delivered orders** with missing approval timestamps were considered inconsistent, as delivered orders are expected to have an approval timestamp. Therefore, these records were removed from the dataset to maintain data quality and consistency.

### Handling Missing Values in `product_category_name`

The `product_category_name` column contained **610 missing values**.

Since these products still exist but their categories are unavailable, removing them would result in unnecessary data loss.

**Action Taken:** Missing values were replaced with **"Unknown"** to preserve all product records while clearly identifying products with unavailable category information.

```python
#filling product_category_name with unknown
products["product_category_name"] = products["product_category_name"].fillna("Unknown")
```

#### product_name_lenght

The `product_name_lenght` column was left unchanged because it is not required for sales analysis and the missing values do not affect business insights.

```python
products["product_name_lenght"].isnull().sum()
```

The `product_description_lenght` column was retained with missing values since it is not used in the analysis and does not impact sales or profitability calculations.

#### product_photos_qty
Missing values in `product_photos_qty` were replaced with **0**, indicating that photo information was unavailable for those products.

```python
products["product_photos_qty"] = products["product_photos_qty"].fillna(0)
```

#### Weight & Dimensions (2 missing)

The product dimension columns (`product_weight_g`, `product_length_cm`, `product_height_cm`, and `product_width_cm`) contained only **2 missing records**.

Since these represented a negligible proportion of the dataset, the corresponding rows were removed to maintain data quality.

```python
products = products.dropna(
    subset=[
        "product_weight_g",
        "product_length_cm",
        "product_height_cm",
        "product_width_cm"
    ]
)
```

The `review_comment_title` and `review_comment_message` columns contained a large number of missing values.

These missing values indicate that many customers provided only a numerical review score without leaving a written title or comment. Since this represents normal customer behavior rather than missing data caused by an error, the review records were retained.

To maintain consistency and simplify text analysis, missing values were replaced with:

- **"No Title"** for `review_comment_title`
- **"No Review"** for `review_comment_message`

This approach preserves all review records while clearly indicating the absence of textual feedback.

## Feature Engineering

Create useful columns.

```python
#Delivery Days
orders["delivery_days"] = (

    orders['order_delivered_customer_date']
    - orders['order_purchase_timestamp']
).dt.days
```

```python
#Purchase Year
orders['purchase_year'] = (
    orders['order_purchase_timestamp'].dt.year
)
```

```python
#purchase month
orders["purchase_month"] = (
    orders['order_purchase_timestamp'].dt.month_name()
)
```

```python
#Purchase Quarter
orders["purchase_quarter"] = (
    orders["order_purchase_timestamp"].dt.quarter
)
```

```python
#Purchase Weekday
orders["purchase_weekday"] = (
    orders['order_purchase_timestamp'].dt.day_name()
)
```

## Translate Product Categories

```python
## Merge English category names.
products = products.merge(
    category,
    on = 'product_category_name',
    how = 'left' 
)
```

```python
products.head()
```

```python
# Merge Orders + Customers
final_dataset = orders.merge(
    customers,
    on='customer_id',
    how='left'
) 
```

```python
# Merge Order Items
final_dataset = final_dataset.merge(
    orders_items,
    on='order_id',
    how='left'
)
```

```python
#Merge Products
final_dataset = final_dataset.merge(
    products,
    on='product_id',
    how='left'
)
```

```python
#Merge Payments
final_dataset = final_dataset.merge(
    payments,
    on ='order_id',
    how ='left'
)
```

```python
# Merge Reviews
#Before merging, remove duplicate reviews (some orders have multiple review records).
reviews = reviews.drop_duplicates(subset='order_id')
```

```python
# Then merge:
final_dataset = final_dataset.merge(
    reviews,
    on='order_id',
    how='left'
)
```

```python
# Merge Sellers
final_dataset = final_dataset.merge(
    sellers,
    on='seller_id',
    how='left'
)
```
